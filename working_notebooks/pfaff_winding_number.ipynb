```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
t = 1.0       # hopping
Delta = 1.0   # p-wave pairing
mu_values = [3.0, 2.0, 1.0]  # Example: trivial (mu>2t) and topological (mu<2t)

k = np.linspace(0, 2*np.pi, 500)  # momentum values

plt.figure(figsize=(8,8))

for mu in mu_values:
    # Kitaev chain Hamiltonian vector
    h_y = 2 * Delta * np.sin(k)
    h_z = -mu - 2 * t * np.cos(k)

    plot = plt.plot(h_z, h_y, label=f'mu = {mu}')

# Plot origin
plt.plot(0,0,'ko',label='Origin (E=0)')
plt.xlabel('$h_z(k) = -\mu - 2t\cos k$')
plt.ylabel('$h_y(k) = 2 \Delta \sin k$')
plt.title('Kitaev chain vector winding in y-z plane')
plt.legend()
plt.grid(False)
plt.axis('equal')

plt.savefig('kitaev_chain_winding.png')
plt.show()
```
